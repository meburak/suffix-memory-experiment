```python
import fasttext
import fasttext.util
fasttext.util.download_model('tr', if_exists='ignore')
```

```python
ft = fasttext.load_model('cc.tr.300.bin')
```

```python
word_list = [
    "MASA", "CAM", "SU", "İYİLİK", "TÜTÜN", "KELİME", "KALEM", "ÖLÇEK",
    "ÇANTA", "AĞAÇ", "PARK", "SARI", "ŞİŞE", "BEYİN", "BARDAK", "MASA",
    "YAZI", "KAĞIT", "BİNA", "KOLAY", "RAKAM", "KÖPEK", "EKRAN", "KALE",
    "TABAK", "TERLİK", "ŞİŞE", "KEDİ", "CÜZDAN", "KAPI", "GRUP", "BÜYÜK",
    "DEFTER", "SPOR", "TOPLU", "ELMA", "KAŞIK", "SEHPA", "KISA", "İLAÇ",
    "KOLTUK", "PARKUR", "TOPRAK", "KALAS", "DOLAP", "KİTAP", "ODA", "HALI",
    "KÜÇÜK", "DUVAR", "SALÇA", "CAM", "ÇİÇEK", "YATAK", "LAMBA", "DÜĞME",
    "KOKU", "FIRIN", "TAHTA", "MARKET", "MEKTUP", "KASA", "HARİTA", "DERGİ",
    "PARA", "ZEMİN", "HALKA", "TERAS", "SIRA", "FIRIN", "KUTU", "BAHÇE",
    "SİLGİ", "KALIP", "ÇORAP"
]
```

```python
ft.get_word_vector("kalıp")
```

```python
word_vec = {}
for word in word_list:
  if word not in  word_vec:
    word_vec[word] = ft.get_word_vector(word)
word_vec
```

```python
def get_counts(input):
  counts = {}
  for w in input:
    if w not in counts:
      counts[w] = 1
    else:
      counts[w] += 1
  return counts
get_counts(word_vec.keys())
```

```python
import pandas as pd
dataframe = pd.DataFrame(data=word_vec)
dataframe
```

```python
dataframe.to_csv("word_vec_fasttext.csv", encoding="utf-8-sig",index=False)
```

```python
import numpy as np
vector_list = [word_vec[word] for word in word_vec]
vec_stack = np.vstack(vector_list)
vec_stack.shape
```

```python
norm_vecs = vec_stack / np.linalg.norm(vec_stack, axis=1)[:, None]#transposed norms
sim_matrix = np.dot(norm_vecs, norm_vecs.T)
```

```python
sim_matrix.shape
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(20, 18))
sns.heatmap(sim_matrix,
            xticklabels=word_vec.keys(),
            yticklabels=word_vec.keys(),
            vmin=0.3, # This "clips" the noise
            vmax=0.8, # This makes strong similarities look brighter
            cmap="magma")
```

```python
word_keys = list(word_vec.keys())
masa_index = word_keys.index("MASA")

# Get similarity scores for 'MASA'
masa_similarities = sim_matrix[masa_index]

# Create a list of (word, similarity) tuples
similarity_scores = []
for i, score in enumerate(masa_similarities):
    similarity_scores.append((word_keys[i], score))

# Sort by similarity score in descending order
similarity_scores.sort(key=lambda x: x[1], reverse=True)

# Display the top 10 most similar words (excluding 'MASA' itself)
print("Most similar words to 'MASA':")
for word, score in similarity_scores[1:11]: # Start from 1 to exclude 'MASA' itself
    print(f"- {word}: {score:.4f}")
```

### Verification of Word-Vector Mapping
Let's verify that the labels match the vector representations by comparing the vector of a specific word from the `word_vec` dictionary with its corresponding vector in the `vec_stack`.

```python
# Pick a word, for example, "CAM"
word_to_check = "CAM"

# Get its vector directly from the word_vec dictionary
vector_from_dict = word_vec[word_to_check]

# Get its index in the word_keys list
index_in_stack = word_keys.index(word_to_check)

# Get its vector from the vec_stack using the index
vector_from_stack = vec_stack[index_in_stack]

# Compare the two vectors
print(f"Vector for '{word_to_check}' from word_vec:\n{vector_from_dict[:5]}...") # Display first 5 elements
print(f"Vector for '{word_to_check}' from vec_stack (at index {index_in_stack}):\n{vector_from_stack[:5]}...") # Display first 5 elements

# Check if they are exactly equal
vectors_match = np.array_equal(vector_from_dict, vector_from_stack)
print(f"\nDo the vectors match? {vectors_match}")
```
