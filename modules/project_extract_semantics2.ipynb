```python
import fasttext
import fasttext.util
fasttext.util.download_model('tr', if_exists='ignore')
```

```python
ft = fasttext.load_model('cc.tr.300.bin')
```

```python
word_list = ['SARI', 'KAĞIT', 'YAZI', 'EKRAN', 'PARK', 'ŞİŞE', 'KOLAY', 'AĞAÇ',
       'KALE', 'KALEM', 'BEYİN', 'ÇANTA', 'ÖLÇEK', 'BARDAK', 'MASA',
       'KÖPEK', 'RAKAM', 'BİNA', 'DEFTER', 'BÜYÜK', 'KAŞIK', 'GRUP',
       'KAPI', 'İLAÇ', 'ELMA', 'KEDİ', 'TABAK', 'KOLTUK', 'CÜZDAN',
       'SPOR', 'KISA', 'TOPLU', 'SEHPA', 'TERLİK', 'KİTAP', 'KALAS',
       'HALI', 'FIRIN', 'LAMBA', 'PARKUR', 'KÜÇÜK', 'KOKU', 'YATAK',
       'TOPRAK', 'ÇİÇEK', 'DÜĞME', 'SALÇA', 'CAM', 'DOLAP', 'ODA',
       'DUVAR', 'KASA', 'DERGİ', 'KUTU', 'SİLGİ', 'MARKET', 'ÇORAP',
       'TERAS', 'KALIP', 'TAHTA', 'HALKA', 'BAHÇE', 'ZEMİN', 'HARİTA',
       'MEKTUP', 'PARA', 'SIRA', 'KABLO', 'TAŞLIK', 'KOLA', 'SULUK']
```

```python
word_vec = {}
for word in word_list:
  if word not in  word_vec:
    word_vec[word] = ft.get_word_vector(word)
word_vec
```

```python
def get_counts(input):
  counts = {}
  for w in input:
    if w not in counts:
      counts[w] = 1
    else:
      counts[w] += 1
  return counts
get_counts(word_vec.keys())
```

```python
import pandas as pd
dataframe = pd.DataFrame(data=word_vec)
dataframe.to_csv("word_vec_fasttext.csv", encoding="utf-8-sig",index=False)
```

```python
import numpy as np
vector_list = [word_vec[word] for word in word_vec]
vec_stack = np.vstack(vector_list)
vec_stack.shape
```

```python
norm_vecs = vec_stack / np.linalg.norm(vec_stack, axis=1)[:, None]#transposed norms
sim_matrix = np.dot(norm_vecs, norm_vecs.T)
```

```python
word_keys = list(word_vec.keys())
similarity_df = pd.DataFrame(sim_matrix, index=word_keys, columns=word_keys)
display(similarity_df.head())
```

```python
similarity_df.to_csv("word_similarity_matrix.csv", encoding="utf-8-sig")
```
